# Machine Learning - Beginner's course

## Supervised Learning vs Unsupervised Learning

**Supervised learning**: needs labelled data to supervise the algorithm when learning from the data. (regression and classification models).

**Unsupervised learning**: no need labelled data that can supervise the algorithm when learning from the data. (clustering and outlier detection models)


### Supervised learning

**Regression:** predicts continuous values (prediction tasks) but **Classification** predicts classification values (decision-making tasks)

Regression learning performances:
- RSS: Residual Sum of Squares
- MSE: Mean Square Error
- RMSE: Root Mean Square Error
- MAE: Mean Absolute Error

Classification learning performances:
- Accuracy: CorrrectPrediction/(CorrectPrediction+IncorrectPrediction)
- Precision: TruePositive/(TruePositive+FalsePositive)
- Recall: TruePositive/(TruePositive+FalseNegative)
- F1-Score: Recall*Precision/(Recall+Precision)

Clustering performances:
- Homogeneity
- Silhouette score
- Completeness

## Training Machine Learning Model

1. Data Preparation (splitting data)
2. Model training (training the model on training data and saving the fitted data)
3. Hyper-parameter tuning (find the optimized parameters where the model performs the best)
4. Prediction
5. Test error Rate: compute the performance metrics using the predictions and the real values of the target variable from your data

## Bias-Variance TradeOff

**Bias of a ML model:** is its inability to capture the true relationship between the data, mathematically it's equal to the difference between the Expectation of model estimation and its true value.

**Variance of a ML model**: is the inconsistency level of model performance when applying it to different data sets.

It is important a ML model that achieves both a low bias and a low variance

## Overfitting and Regularization

**Overfitting:** occurs when the model performs well in training (low training rate error) while perform worst in test data (high testing error). High model variance and low bias

### How to fix overfitting ?
1. Reduce complexity model
2. Collect more data
3. Use resampling technique (CV)
4. Early stopping
5. Ensemble methods
6. Dropout

**Regularization:** is a method that shrinks some of the estimated coefficients toward zero, to penalize unimportant variables for increasing the variance of the model:

- Ridge regression based on L2 norm (solves overfitting, lower model variance, computationally cheap/ low interpretability)
- Lasso regression based on L1 norm (solves overfitting, easy to understand, high interpretability, high interpretability, feature selection / higher variance than Ridge)
- Dropout in NN

## LINEAR REGRESSION MODEL

**Dependant variable:** is the variable that is being measured or tested. It is called the dependant variable because it is thought to depend on independant variable.

**Independant variable**: is the variable that is being manipulated or controlled in the experiment, and it is believed to have an affect on the dependant variable

**Linear regression:** is a statiscal method that can help to model the impact of a unit change in a variable (independant variable) on the values of another target variable, (dependant variable) when the relationship between two variables is linear.

**OLS(Ordinary Least Squares):** is an estimation technique for estimating unknown parameters in a linear regression model to predict the response/dependant variable.

Linear regression assumptions:

- Linearity
- Random Sample
- Exogeneity
- Homoscedasticity
- Perfect MultiCollinearity

**PROS:**

- Simple model
- Computationally efficient
- HIgh interpretability
- Able to handle missing data

**CONS:**

- Overly-simplistic
- MAny assumptions
- Assumes linearity
- Prone to outliers

## LOGISTIC REGRESSION

**Logistic Regression:** is a supervised learning technique that model conditional probability of an event occuring or observation belonging to a certain class, given a datasetof independant variables.

**MLE (Maximum Likelihood Estimation)** is used to estimate coefficients and parameters in logistic regression

### PROS: ###
 - Simple model
 - Low bias
 - Low variance
 - Provides probability

### CONS: ###
- Unable to model
- Unstable when classes well separable
- Unstable when > 2 classes
