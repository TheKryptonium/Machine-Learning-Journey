## PRINCPIPAL COMPONENT ANALYSIS (PCA)

**PCA** is the method for dimensionality reduction that is used to reduce to variables or the dimensions of the data, by transformaing large set of features into smaller ones that contains most of the information.

+ Principal components are unit vectors or new variables that are constructed as linear combinations of the initial variables
+ These combinations are done in such a way that the principal components are uncorrelated and most of the information within the initial variables is compressed inot the first components.
+ We are projecting each data point onto only the first few principal components to obtain lower-dimensional data while preserving as mush of data's variation as possible

#### PCA Algorithm

1- Data Preprocessing (Data Standardization): X_scaled = (X-mean)/sd

2- Compute the Covariance Matrix

3- Compute the eigenvectors and eigenvalues

4- Rearrange the eigenvectors and eigenvalues

5- Compute the cumulative energy content for each eigenvector

6- Select a subset of eigenvectors as basis vectors

7- Project the data onto the new basis

### Bagging 

Also called bootstrap aggregation. The intuition in bagging: 

- Divide data in samples (it's possible to have samples without replacement)
- Feed the model with a subset of datas
- Train the model based on this
- You repeat the steps 2 and 3
- And the end you obtain many models, in case of classification the majority indicates the prediction and in regression case the mean is computed

Bagging helps reducing the variance. The based models in this process are usually high variance and low bias models, their combinations provide to obtain a low bias and low variance model.

Sample method (row sampling)

Random Forest is one the most powerful bagging algorithm is a combination of decision trees + bagging + feature bagging (column sampling), the based models are exclusively decisions trees models. 

### Boosting

It's the opposite of bagging in trade off term, here we have high bias and low variance base models additively combined. It helps to reduce the bias. The intuition behind boosting:

- Train the data
- Compute the cost/ residual error : ground truth - predicted values.

The different boosting algorithms:

- Adaptative Boosting (AdaBoost): not very used in production
- Extra Gradient Boosting (XGBoost)
- Gradient Boosting

#### Gradient Boosting

Intuition:

- The inputs are the training data, a differentiable loss function and number of iterations (number of the base learners)
- Initialize the  model with a constant value
- Iterate :
    - Compute the so-called residuals
    - Fit a base learner closed under scaling to pseudo residuals (that means to train it using the training set)
    - Compute the multiplier by solving a one dimensional optimizer equation
    - Update the model
- Output (big model)

Generally due to additive residual, the model starts to overfit (variance gets higher while bias goes lower), that justify why we use regularization by shrinkage


#### Extra Gradient Boosting 

It's a better version of adaptative boosting (gradient boosting + column sampling + row sampling), it's very fast

### Stacking 

**Stacking:** is an ensemble learning technique to combine multiple classification models via meta-calssifier. The individual classification models are trained based on the complete training set, then, the meta-classifier is fitted based on the outputs - meta features of the individual classification models in the ensemble. The meta-classifier can either be trained on the predicted class labels or probabilities from the ensemble.

The base learners are quite good and well fine tuned, training data is subvided in subsets which provide to train the base learners.

Intuition:

- Input : trainiing dataset and output: an ensemble classifier
- Learn first level classifiers
- Iterate accoriding to the number of base learners:
    - Learn a base classifier based on dataset
- Construct new datasets from the training dataset
- Learn a second-level classifier
- Repeat the process 

## UNSUPERVISED LEARNING

**Unsupervised learning:** is a machine learning subfield where we have a dataset but without a supervisor (no labelled data). There's 2 types of unsupervised learning problems: continuous and classification problems (clustering/segmentation).

- **Segmentation/clustering:** recommandations engine (content filtering, collaborative filtering), customers habits, sequence analysis, social network analysis, image segmentation (object detection, classification), sentiment analysis (NLP), anomaly detection

Clustering = segementation, it's an unsupervised ML technique where elements are grouped in such that these elements should be similar to each group element and different from another cluster group.

- Inter cluster: it's the distance between two clusters
- Intra cluster: it's the distance between two points included in a defined cluster

Evaluation techniques in clustering:

- Dunn Index: D = (max d(i, j))/(max d(k)), the numerator represents the max intercluster distance, the denominator represents the max intracluster distance, the higher D is, the better the model is
- Davies Bouldin Index: DB = (1/n)sum(max(sigmaI, sigmaJ)/d(ci, cj))

The types of clustering: 

- Partitional based clustering
- Hierarchical clustering
- Well separated
- Center based
- Density Based (DBSCAN)

### K-MEANS

Also called LLoyd's algorithm/K-Means clustering algorithm.

Intuition:

- Initialize the centroids k
- Repeat:
    - Assign all the clusters
    - Taking the average and recompute the centroid position until your centroids are not changing

Evaluation techniques:

- Sum Square Error: based on euclidian distance (minimize the intraclustering distance)

Limitations:

- The centroid is initialized randomly which increases time complexity.
- Sensitive to outliers
- Solutions:
    - KMeans++: select centroids based on multiple runs 
    - Elbow method: select the number of k

### Hierarchical clustering

**Hierarchical clustering:** indicates a hierarchy of clusters. There are two types of hierarchical clustering:

- Agglomerative clustering: down to up approach
    - Compute the proximity matrix
    - Repeat (untitl all the clusters are covered):
        - Merge two clusters
        - Update the proximity matrix. Methods used to measure similarities:
            - MIN
            - MAX
            - GROUP average

- Divisive clustering: up to down approach

- It's also very sensitive to outliers
