```python
root = "https://www.football-data.co.uk/mmz4281/"
```

```python
def full_season(year):
    return f"{year}{year+1}"
```

```python
leagues = ["E0", "E1", "E2", "E3"]
frames = []

for league in leagues:
    for season in range(15,26):
        df = pd.read_csv(f"{root}{full_season(season)}/{league}.csv")
        df.insert(1, "season", season)
        frames.append(df)
```

```python
len(frames)
```

```python
frames[0].tail()
```

```python
dict_countries = {
    "Spanish La Liga": "SP1", "Spanish Segunda division": "SP2",
    "German BundesLiga": "D1",
    "English Premier League": "E0", "English League 1": "E1", "English League 2": "E2"
}
```

```python
dict_historical_data = {}

frames = []

for league in dict_countries:
    frames = []
    for season in range(15,26):
        df = pd.read_csv(f"{root}{full_season(season)}/{dict_countries[league]}.csv")
        df.insert(1, "season", f"{season}-{season+1}")
        frames.append(df)
    df_league = pd.concat(frames)
    dict_historical_data[league] = df_league
```

# DATA CLEANING

## Loading dataset

```python
df_netflix = pd.read_csv("./data/netflix_titles.csv")
df_netflix
```

```python
df_netflix.shape
```

## Identifying missing data

```python
#Number of rows missing in each column
df_netflix.isnull().sum().sort_values(ascending=False)
```

```python
percentage={}
for column in df_netflix.columns:
    percentage[column] = df_netflix[column].isnull().mean()*100
percentage
```

## Dealing with Missing data

```python
#Drop column
df_netflix.drop("director", axis=1)
#Drop row
directorless = df_netflix[df_netflix["director"].isnull()].index
df_netflix.drop(directorless, axis=0)
#isnull()
df_netflix[~(df_netflix["date_added"].isnull())]
#dropna
df_netflix.dropna(subset=["director"])
```

```python
df_netflix["director"].ffill() #Forward fill
```

```python
df_netflix["director"].bfill() #Backward fill
```

```python
df_netflix.fillna({"duration":"0 min"}, inplace=True)
df_netflix.ffill()
df_netflix["duration"].isnull().sum()
```

```python
df_movie = df_netflix[df_netflix["type"]=="Movie"]
df_movie["duration"].isnull().sum()
df_movie["minute"]=df_movie["duration"].str.split(expand=True)[0].astype(int)
```

```python
df_movie.dtypes
```

```python
df_movie["date_added"].str.split(",",expand=True)[1]
df_movie["date_added"].str.extract("(\d{4})")
```

```python
df_movie["minute"].plot(kind="hist", bins=10)
```

```python
df_movie["minute"].value_counts(bins=10).sort_index()
```

```python
df_movie["rating"].value_counts().plot(kind="bar")
```

```python
df_movie.loc[:,"title"] = df_movie["title"].str.upper()
df_movie.loc[:,"title"] = df_movie["title"].str.strip()
df_movie
```
