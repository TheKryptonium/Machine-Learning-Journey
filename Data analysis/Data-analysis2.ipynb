## NUMPY

Python(pure) is very slow to process large amount of datam it's almost inefficient for deeper level of processing, Numpy solves that. Tools like pandas and matplotlib relies on numpy.
A Python list for example uses a large amount of storage, but by using numpy you can optimize that storage and make calculations on data much more faster.
Python is a higher level programming language, all elements in Python (number, string,...) are all objects and take to much storage.
Numpy we can be very precise about the amount of storage we want to use, numpy is array-oriented which is almost like lists, tuples,... (which are not really efficient and take sometimes a large amount of space).
