##  What is Machine Learning ?

Machine Learning is an Artificial Intelligence domain where we extract patterns from the data and analyze data and make intelligent predictions on the new data according to the pattern the machine has learnt.

AI problems can be solved if we get right set of features and make a mapping from feature to desired output, say for an example you want to detect a cat in an image, so you'll hand desugn right set of features like ears, nose which is very time consuming task and getting the right features is also challenging for many tasks around us.

$$\text{Input}\longrightarrow\text{Feature extraction}\longrightarrow\text{Classification}\longrightarrow\text{Output}$$

- One solution to this problem is making machine learning not only learn mapping frim representation to y but also learn representation itself
- The model takes the raw data as input, performs feature extraction by learning the feature representations and learns the parameters to perform the necessary task such as classification


## What's Deep Learning ?

Deep learning methods aim at learning feature hierarchies with features from higher levels of the hierarchy formed by the composition of lower level features. Automatically learning features at multiple levels of abstraction allow a system to learn complex functions mapping the input to the output directly from data, without depending completely on human-crafted features.

## Introduction to Neural Networks

In biological NN, the dendrites are the input terminals, the axon (output wire) and the axon terminals (output terminals). Neurons are connected to one another wiht the use of axons and dendrites are the connecting regions between axons and dendrites referred as synapses.

Information which you have $(x_1, x_2, x_3, x_4)$ received in dendrites and the information is weighted by weights $(w_1, w_2, w_3, w_4)$, it determines the effects of the input $(w_1x_1, w_2x_2, w_3x_3, w_4x_4)$.
The weighted information are aggregated in the nucleus as weighted sum $z=\sum_{i=1}^n w_i x_i + b$ where $b$ is the bias term and then we apply some non-linear function $y=h(z)$ and then it is sent for further processing to axon $y$ and it reaches to destination or sent to other neurons via dendrites.
The basic unit of NN is inspired by traditional ML, as we put together such basic units to makemore powerful models.

## The Perceptron

The perceptron is the simplest neural network.
- Given $X=[x_1, x_2, x_3,...,x_d], y\in{0,1}$
- Weights: $W=[w_1,w_2,w_3,...,w_d]$ where Learning occurs by changing the weights and the goal of changing the weights is to modify the computed function to make the predictions more correct in future iterations
- Aggregate the weighted information: $z=\sum_{i=1}^n w_ix_i+b$ and then apply the activation function $\sigma(z)=\frac{1}{1+e^{-z}}$ or non linearity before sending ot its destination.
- Bias allows us to shift the activation function by adding a constant tothe input
- $b$ shifts the graph and allows us to represent more complex situations

## What is Learning ?

- Learning happens by changing the values of our weights and in every iteration of it, loss value should decrease. In brief, we will use our batch gradient desciend algorithm ot minimize our loss function or it will help us to reach global minima in our function $J(\theta)$. With: $$J(\theta)=\frac{1}{n}\sum_{i=1}^{n}(h(x)-y)^2,$$ $$\quad \theta \text{ is the vector containing all the weights},$$ $$y \text{ is the ground truth},$$ $$h(x)=\theta.X^T \text{ where h is the activation function, X is the features vector}$$
- In an iteration gradient descent tweaks the weight and checks if the loss value or error term decreases, if yes then we update the value of our weight and keep on doing this step until and unless we converge or our loss is minimized.

Here are the steps:
- Compute the loss function
- Optimization by resolving the argmin of the loss function
- Solve optimization by using a learning rate until convergence
    - Initialize weights
    - Repeat until convergence: $w_{new}=w_{old}-\alpha[\nabla_wJ]w_{old}$
    -  $\nabla_wJ=(\frac{\partial J}{\partial w_1},\frac{\partial J}{\partial w_2},...,\frac{\partial J}{\partial w_n})^T$
