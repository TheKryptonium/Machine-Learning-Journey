## What's Deep Learning ?

**Machine Learning:** is a technique which consists in teaching computers to recognize patterns in data.
**Deep Learning:** is a field of ML that learns features and tasks directly from data. Inputs are run through neural networks which have hidden layers.

## Why Deep Learning ?

In traditional ML, there's a big need of human expertize, labelling... In Deep Learning, the computer learns features just by raw data without a big dependance in human intervention.

- Data is much more prevalent (Big Data)
- Improved Hardware Architecture
- New Software Architecture 

## Neural Networks

There are inspired from the human brain. An entity in a NN architecture is called: **neuron** exactly as the human brain.
NN take in data as input, train themselves to understand patterns in data, output useful predictions.

A simple NN architecture contains:
- An input layer
- A hidden layer
- An output layer

## The Learning Process of a Neural Network

There are two main propagations:
-  Forward Propagation: from the input to the output
-  Backward Propagation: from the output layer to the input layer, it's the reason why NN are so powerful, it's very helpful to compute loss functions

Some keywords:
- Weight: is an indicator which tells how important is a neuron
- Bias: allows for the shifting of the activation function to the right or the left
- Loss functions: help quantify the deviation from the expected output

Learning process algorihtm:
- Random initialization
- Feed input data to network
- Forward propagation (from the input to output by applying weight,  bias and activation function)
- Back propagation (compute the loss function, go backwards and adjust initial weights and biases, values adjusted to better fit prediction model)
- Iterate the previous steps
- Now it's able to make predictions

## Most common terminlogy used in Deep Learning today

### Activation functions

**Activation function:** a non-linear function, decide whether a neuron can
contribute to the next layer.

Some activation functions:
- Linear
- Unit step function
- Sigmoid ($\sigma (x)$)
- ReLU
- Modified ReLU
- Hyperbolic tangent ($\tanh (x)$)

Which activation function to use:
- Binary classification problems: sigmoid
- If you're unsure: ReLU (or modified ReLU)

Why non-linear activation functions:
Non linear functions are able to describe complex patterns 


### Loss functions

**Loss function:** Quantify the deviation of the predicted output by the neural network to the expected output.
- Regression: Squared Error, Huber Loss
- BInary classification: Binary Cross-Entropy, Hinge Loss
- Multi-Class classification: Multi-Class Cross-Entropy, Kullback Divergence

### Optimizers

During the training porcess, we adjust the parameters to minimize the loss function and make our model as optimized as possible. **Optimizers** tie together the loss function and model parameters by updating the network based on the outptut of the loss function.

The loss functions direct the optimizer

The mainly used optimizers:
- **Gradient descent:** iterative algorithm that starts off a random point on the loss function and travels down its slope in steps until it reaches the lowest point of the function. It's the most popular optimizer, fast, robust and flexible. Here is its algorithm:
    - Calculate what a small change in each individual weight would do to the loss function
    - Adjust each parameter based on its gradient (take a small step in the determined direction)
    - Repeat steps 1 and 2 until the loss function is as low as possible
- **Stochastic Gradient Descent:** is an implementation of Gradient Descent that uses batches on each pass, uses momentum to accumulate gradients, less intensive computationally. Back propagation is basically the gradient descent implemented in a NN
- **Adagrad:** adapts learning rate to individual features, some weights will have different learning rates, ideal for sparse datasets with many input examples missing, learning rate tends to get small with time
- **RMSprop:** specialized version of adagrad, accumulates gradients in a fixed window, similar to adaprop
- **Adam:** (Adaptive Moment Estimation), uses the concept of momentum, pretty widespread today

### Parameters and hyperparameters:

**Model parameters:** variables internal to the neural network. Value can be estimated right from the data. They are required by the model to make predictions, values define the skill of the model, estimated directly from data, not set manually, saved as part of the learned model (weights, biases..). When you save your model, you essentially save your model parameters.

**Model Hyperparameters:** configuration external to the neural network. Value cannot be estimated right from the data. no clear-cut way to find the best value, when a DL algorihtm is tuned, you are really tuning the hyperparameters, if you have to manually specify a parameter, it is a hyperparameter (Learning rate ($lr$), $C$ and $\sigma$ in SVMs)

Model parameters are estimated from the data, model hyperparameters can't be estimated from the data. Model hyperparameters are often referred to as parameters because they are the parts of the machine learning that must be set manually and tuned.

### Epochs, Batches, Batch Sizes & Iterations
Need these terms only if dataset is large and break down the dataset into smaller "chunks" and feed those chunks to the NN one-by-one

**Epochs:** when the entire dataset is passed forward and backward through the neural network only ONCE. We use multiple epochs to help our model generalize better(example: you must repeat a song a couple of times to memorize the lyrics)

**Batch & Batch Size:** we divide large datasets into smaller batches and feed those batches into the neural network. The batch size: is the total number of training examples in a batch

**Iterations:** number of batches needed to complete one epoch, alternatively,$\quad \text{number of batches = number of iterations for one epoch}$

## Types of Learning

- **Supervised Learning:** algorithms designed to learn by example, models are trained on well-labelled data. The objective is to predict the correct label for unseen data. It can be split into two categories: classification(predicts a category or a class) and regression (predicts continuous values).
    - During training, SL algorithm searches for patterns that correlate with the desired output
    - After training, it takes in unseen inputs and determine which label to classify it to

- **Unsupervised Learning:** focuses on data features that a human observer might not pick up on. Categories of UL: clustering, association

- **Reinforcement Learning:** enables an agent to learn in an interactive environment by trial and error based on feedback from its own actions and exepriences, uses rewards and punishments for positive and negative behavior. The goal is to find a suitable model that would maximize the total cumulative reward

## Regularization

There are mainly two probems in learning:
- Underfitting: when the model performs bad in train and test set
- Overfitting: when the model has a high bias and a low variance, the model performs very well in train set but worse in test set. It's the main learning problem in the data science domain.

There are several methods to handle and tackle overfitting:
- Dropout: every iteration selects some node randomly and removes them and their connections, if the model performs better, it's adopted
- Augmentation: more are the data, better the model is, one popular technique is to use more fake data by applying transformations on the existing dataset to get synthesize more data
- Early stopping: training error decreases steadily but validation error increases after a certain point

## Neural Network Architectures

### Fully connected feed forward neural network
Each neuron is connected to every subsequent layer with no backward connections, \$$\text{more neurons = larger computational resources}$$. It takes a fixed size of inputs and gives a fixed size of outputs
### Reccurent Neural Networks
FFNN are unable to process sequential data (data in a sequence), don't share parameters across time. Sharing parameters gives the network the ability to look for a given feature everywhere in the sequence, rather than in just a certain area, RNN deal with variable length sequences, maintain sequence order, keep track of long-term dependecies, share parameters across the sequence. The RNN uses a feedback loop in the hidden layers, decisions are influenced on what they have learned in the past.
Training process of an RNN:
- Back propagation algorithm
- Backdrop applied for every sequence data point
- Backpropagation through time (BTT)
### LSTM and GRNN
LTSM: Long Short Term Memomry (Update gate, reset gate, forget gate)
GRNN: Gated RNN (update gate, reset gate)
There are capable of learning long-term dependencies using gates
### Convolutional Neural Networks (CNNs)
Inspired by the organization of neurons in the visual cortex of the human brain, good for processing data like images, audio and video. Hidden layers in a CNN: 
- Convolutional layers
- Pooling layers
- Fully-connected layers
- Normalization layers

The input is generally a 2D array and the output a 1D array. 
- **Convolution:** allows us to extract visual features in chunks.
- **Pooling:** reduces the number of neurons necessary in subseqent layers

The CNN process algorihtm:
- Convolve the image
- Pool the results
- Repeat
- Add few layers to help classify the image
- Prediction in the output layer

## Creating a Deep Learning Model

### Gathering Data

Picking the right data is key, bad data imply a bad model, make assumptions about data you need. There's no size-fit-all.
$$\text{amount of data you need = 10}\times \text{No. of model parameters} $$
Quality matters too, there's no reason to have a large amount of data if the quality is poor.

### Preprocessing step

Splitting the dataset into subsets (training set, validation and testing), it can't be splitted randomly.
$$\text{few hyperparameters} \longrightarrow \text{small validation set}$$
$$\text{many hyperparameters} \longrightarrow \text{large validation set}$$
$$\text{no hyperparameters} \longrightarrow \text{no validation set required}$$

Cross-validation is a cool technique to avoid overfitting. Time-splits are best suited for large dataset.
It's very important to deal with missing data (eliminating features with missing values or imputing the missing values)
For  very large datasets, it's important to use sample to gain in time.
$$\text{Example weight = Original Weight} \times \text{DownSampling Factor}$$
The advantages of sampling: faster convergence, reduced disk space, dataset is in similar ratio.

Feature Scaling is also a crucial step in deep learing: normalization, standardization


### Training the model

$$\text{Feed data} \longrightarrow \text{Forward Propagation} \longrightarrow \text{Loss Function} \longrightarrow \text{Backpropagation}$$

### Evaluation

Test model on the validation set, meant to be representative of the model of the real world

### Optimization

- Hyperparameter tuning: increasing the number of epochs, adjust learning rate, initial conditions play a large role in determining your model's outcome
- Adressing overfitting: getting more data, reducing models size, regularization (L1 and L2 regularization)
- Data augmentation: good way of increasing dataset artificially (flipping, blurring, zooming)
- Dropout: randomly drops out some neurons, reduce co-dependency if neurons
